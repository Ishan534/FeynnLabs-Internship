LOADING ALL THE RECIEPTS THAT'S BEEEN OCR USING PADDLEOCR 

```python
import os
import pandas as pd

# Folder containing ground truth text files
gt_folder = 'DataSet/train/box'

# Create a list to store results
results = []

# Loop through all text files
for file in os.listdir(gt_folder):
    if file.endswith('.txt'):
        file_path = os.path.join(gt_folder, file)

        clean_lines = []
        
        with open(file_path, "r", encoding="utf-8") as f:
            for line in f:
                parts = line.strip().split(",")
                if len(parts) > 8:
                    text = ",".join(parts[8:])  # join everything after the first 8 numbers
                    clean_lines.append(text)
        
        # Join lines into a single receipt text
        clean_text = "\n".join(clean_lines)

        # Replace .txt with .jpg (or .png depending on your case)
        image_name = file.replace('.txt', '.jpg')

        results.append({
            'FileName': image_name,
            'Text': clean_text,
            'Confidence': 1.0
        })

# Create DataFrame
df = pd.DataFrame(results)
```

```python
def remove_dollar(s):
    
    if not s:  # handles empty strings or None
        return "0"
    s = s.replace(",", "")
    if s[0] == '$':
        return s[1:].strip()
    if s[0] == 'R' and s[1] == 'M':
        return s[2:].strip()
    return s.strip()  # return original if no '$' at start
cleaned_text = []
cleaned_total = []

for i in range(626):
    text = df['Text'][i]
    cleaned_text.append(text[14:])  # remove first 14 chars

    total_value = df['total'][i]
    cleaned_total.append(remove_dollar(total_value))  # remove '$' if present

df['Text'] = pd.Series(cleaned_text)
df['total'] = pd.Series(cleaned_total)
```
